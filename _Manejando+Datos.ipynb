# Indexando Time Series 

Aqui hacemos 3 operaciones a la vez:
- 1. Leemos el archivo de datos
- 2. Convierte la columna "Time" a formato datetime (actualmente en String).
- 3. Convierte la columna "Time" en el Indice del DataFrame

```
ruta_del_archivo = "------ ruta al archivo .csv ------------"
```

```
data1 = pd.read_csv(ruta_del_archivo, parse_dates=['Time'], index_col='Time')
```

```
data1.head()
```

```
type(data1.index)
```

```
data1.index = data1.index.strftime('%Y-%m-%d %H:%M')
```

```
type(data1.index)
```

# Resampling *(Cambiando la frecuencia de los datos)*

El Resampling funciona como una forma de 'agrupación'. 

Por ejemplo: Si resampleamos por día 'D', significa que agrupará los datos del Dataframe por días, esto lo podemos verificar usando el método de 'lista' (como en los ejemplos a continuación)

```
data1.resample("D")
```

```
data1_por_dia = data1.resample("D") 
```

```
data1_por_dia = data1.resample("D")
```

```
list(data1_por_dia)
```

```
list(data1.resample("D"))[4][1]
```

Como el Resampling en realidad está agrupando valores, necesitamos tener un valor que represente el nuevo muestreo.

Por ejemplo: Tenemos dataframe inicial que esta en horas, por lo que para convertir todas las horas dentro de un solo valor que sera el día, necesitamos un valor que represente ese día, que podria ser el *** first *** valor del grupo, o el *** last *** o la *** mean ***, *** median *** , *** sum ***, etc.

- * mean * = El valor promedio
- * median * = El valor intermedio

```
data1.resample('D').first() 
```

```
data1.resample('D').mean() 
```

```
data1.resample('2H').first()
```

```
data1.resample('W').mean()
```

```
# Aqui seleccionamos un dia especifico de la semana
data1.resample('W-Wed').mean()
```

```
# "M" devolvera valores mensuales (empezando desde el final de cada mes)
data1.resample('M').median()
```

```
# "MS" devolvera valores mensuales (empezando desde el inicio ("S") de cada mes)
data1.resample('MS').mean()
```

```
# Devolvera solamente the 15vo dia de cada mes ("M") (empezando desde el inicio ("S") de cada mes)
data1.resample('MS', loffset='14D').mean()
```

```
# Resampling anual - Tomando el ULTIMO dia del año
data1.resample('Y').sum()
```

```
# Resampling anual - Tomando el PRIMER dia del año
data1.resample('YS').sum()
```

```
# Resampling TRIMESTRAL (Quarterly)
data1.resample('Q').mean()
```

```
# Desde que mes queremos contar los trimestres
data1.resample('Q-Feb').mean()
```

```
import numpy as np
import pandas as pd

import yfinance as yf

# To plot the data
import matplotlib.pyplot as plt

# To better style the plot
import matplotlib.style
plt.style.use('ggplot')
```

```
stocks = ['GOOG', 'SNAP', 'FB', 'AMZN']
start_date = '2010-01-01'
end_date = '2020-01-01'
```

```
precio_de_cierre = pd.DataFrame()
```

```
for ticker in stocks:
  precio_de_cierre[ticker] = yf.download(ticker, start_date, end_date)['Adj Close']
```

```
precio_de_cierre
```

# Llenando valores NaN 

```
precio_de_cierre.fillna(0) # LLenara todos los espacion NaN con '0'
```

```
# 'bfill' (back-fill): 
# LLenara todos los espacios NaN con el primer valor q encuentre pero haceia atras
precio_de_cierre.fillna(method='bfill') 
```

```
for ticker in stocks:
  precio_de_cierre[ticker] = yf.download(ticker, start_date, end_date)['Adj Close']
```

```
precio_de_cierre
```

```
# 'bfill' (back-fill): 
# LLenara todos los espacios NaN con el primer valor que encuentre pero haceia atras
# axis=0: ROWS (default)
# axis=1: COLUMNS
precio_de_cierre.fillna(method='bfill', axis=1) 
# p_cierre.fillna(method='bfill', axis=0) 
```

```
precio_de_cierre.fillna(method='ffill', axis=1) 
```

```
# Eliminando valores NaN (drop)
# axis=0: Eliminara TODAS las filas (rows) que contengan valores NaN 
# axis=1: Eliminara TODAS las columnnas que contengan valores NaN 
precio_de_cierre.dropna(axis=1)
```

**NOTA:** *En lo que respecta a backtesting, es mas conveniente usar bfill (back-fill), en vez de '0', asi no sera un cambio brusco, que si usaramos '0'*

```
# Veremos que el Dataframe NO cambió porque el parámetro 'inplace' estaba como Falso (predeterminado)
precio_de_cierre
```

```
# Para realizar cambios reales, tenemos que establecer el parámetro 'inplace' como True
precio_de_cierre.fillna(method='bfill', inplace=True) 
```
